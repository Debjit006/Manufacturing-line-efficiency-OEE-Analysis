```python
missing_values=df.isnull().sum()
```

```python
missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

missing_report = pd.DataFrame({
    'Missing Values': missing_values,
    'Percentage': missing_percentage
})

missing_report
```

```python
df['defect_reason'] = df['defect_reason'].fillna('Not Recorded')
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
print("Duplicate Record IDs:",
      df['record_id'].duplicated().sum())
```

```python
df = df.drop_duplicates(
    subset=['record_id'],
    keep='first'
)
```

```python
#Convert data types
df['production_date'] = pd.to_datetime(
    df['production_date'],
    errors='coerce'
)

df['production_start_time'] = pd.to_datetime(
    df['production_start_time'],
    errors='coerce'
)

df['production_end_time'] = pd.to_datetime(
    df['production_end_time'],
    errors='coerce'
)
```

```python
df.dtypes
```

```python
text_columns = df.select_dtypes(
    include='object'
).columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
numeric_columns = [
    'planned_production_time_minutes',
    'downtime_minutes',
    'minor_stop_minutes',
    'total_count',
    'reject_count',
    'good_count',
    'ideal_cycle_time_seconds'
]

df[numeric_columns].describe()
```

```python
for col in numeric_columns:
    print(col, "Negative values:",
          (df[col] < 0).sum())
```

```python
print("Rejects greater than total:",
      (df['reject_count'] > df['total_count']).sum())

print("Good + Reject mismatch:",
      (df['good_count'] + df['reject_count']
       != df['total_count']).sum())

print("Downtime greater than planned time:",
      (df['downtime_minutes'] >
       df['planned_production_time_minutes']).sum())
```

```python
df['production_duration_minutes'] = (
    df['production_end_time'] -
    df['production_start_time']
).dt.total_seconds() / 60

df[['production_start_time',
    'production_end_time',
    'production_duration_minutes']].head()
```

```python
invalid_time = (
    df['production_end_time'] <=
    df['production_start_time']
)

print("Invalid time records:", invalid_time.sum())
```

```python
minor_stop_issues = df[
    df['minor_stop_minutes'] >
    df['downtime_minutes']
]

print("Records requiring review:",
      len(minor_stop_issues))
```

```python
quality_report = pd.DataFrame({
    'Column': df.columns,
    'Data Type': df.dtypes.astype(str).values,
    'Missing Values': df.isnull().sum().values,
    'Unique Values': df.nunique().values
})

quality_report
```
